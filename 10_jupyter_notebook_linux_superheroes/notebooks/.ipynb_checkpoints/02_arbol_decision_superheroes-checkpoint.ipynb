# 🌳 Árbol de Decisión — Regresión de Poder de Superhéroes

Análisis comparativo y sistemático usando `DecisionTreeRegressor` entrenado con tres
particiones distintas **(80/20, 70/30, 60/40)** para demostrar la robustez del modelo predictivo.

**Variable predictora (X):** `fuerza`
**Variable objetivo (Y):** `poder`

Pipeline PostgreSQL — datos de 13 superhéroes, 3 535 registros.
Del notebook 01 sabemos que `fuerza` es el predictor con mayor correlación individual
con `poder` (r = 0.395), por eso se usa como única variable predictora en este árbol.

### CELDA 1 —  Configuración Inicial — Importaciones y Estilo Visual

Importa todas las librerías necesarias y establece la conexión a PostgreSQL
reutilizando `scripts/database.py`. Misma configuración visual del notebook 01.

```python
# ── CELDA 1: Importaciones ────────────────────────────────────
import sys
import os
sys.path.insert(0, '..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm

from sklearn.tree import DecisionTreeRegressor, plot_tree, export_text
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score,
)

from scripts.database import engine  # reutiliza el motor del proyecto

# Estilo visual consistente con el notebook 01
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams.update({'figure.dpi': 120, 'font.size': 11})

print("✅ Librerías importadas correctamente")
print(f"   pandas      {pd.__version__}")
print(f"   numpy       {np.__version__}")
print(f"   seaborn     {sns.__version__}")
```

### CELDA 2 — Carga de Datos desde PostgreSQL

Reutilizamos el mismo QUERY del notebook 01, seleccionando únicamente
`fuerza` (X) y `poder` (Y). Se aplica la misma estrategia de imputación
de la Celda 2.5: primero por media del propio personaje, luego media global.

```python
# ── CELDA 2: Carga de datos desde PostgreSQL ─────────────────
QUERY = """
SELECT
    s.nombre AS superheroe,
    p.fuerza,
    p.poder
FROM powerstats p
JOIN superheroes s ON p.superhero_id = s.id
ORDER BY p.fecha_extraccion
"""
df = pd.read_sql(QUERY, engine)

print(f"✅ Datos cargados desde PostgreSQL")
print(f"   Filas    : {df.shape[0]:,}")
print(f"   Columnas : {df.shape[1]}")
print(f"\n📋 Valores nulos:")
print(df.isnull().sum())
print(f"\n📋 Primeras filas:")
df.head()
```

### CELDA 3 — Preparación de Variables (X / Y)

Se construyen los vectores necesarios para el modelo de Machine Learning:

- **X (features):** conjunto de variables predictoras que describen las habilidades del superhéroe (`fuerza`).
- **y (target):** variable objetivo `poder`, que se busca predecir.

> Los datos se convierten en arreglos NumPy (`.values`) para ser compatibles con scikit-learn.

```python
# ── CELDA 3: Limpieza e imputación + vectores X / y ──────────
# Misma estrategia de imputación que la Celda 2.5 del notebook 01
df = df.copy()

for col in ['fuerza', 'poder']:
    df[col] = df.groupby('superheroe')[col].transform(
        lambda x: x.fillna(x.mean())
    )

# Respaldo: media global
df[['fuerza', 'poder']] = df[['fuerza', 'poder']].fillna(
    df[['fuerza', 'poder']].mean()
)

# Vectores para scikit-learn
X = df[['fuerza']].values   # shape: (3535, 1)
y = df['poder'].values       # shape: (3535,)

print(f"✅ Datos limpios y vectores preparados")
print(f"   X (fuerza) — shape : {X.shape}  |  rango : [{X.min():.1f}, {X.max():.1f}]")
print(f"   y (poder)  — shape : {y.shape}  |  rango : [{y.min():.1f}, {y.max():.1f}]")
print(f"\n📊 Estadísticas descriptivas:")
df[['fuerza', 'poder']].describe().round(2)
```

### CELDA 4 — Bloque de Lógica — Función de Evaluación por Split

El corazón del notebook. Una sola función encapsula todo el proceso para cualquier
tamaño de split, evitando duplicación de código y garantizando consistencia entre
las tres particiones.

**¿Por qué `random_state=42`?**
Garantiza que la división aleatoria sea reproducible. Sin este parámetro, cada
ejecución produciría splits diferentes e imposibilitaría la comparación entre sesiones.

Genera **dos gráficos por split**:
- **Gráfico 1:** Residuos vs Valores Predichos (+ curva LOWESS)
- **Gráfico 2:** Predicción vs Real (+ diagonal ideal)

```python
# ── CELDA 4: Función de evaluación reutilizable ──────────────
import os as _os

_GRAFICAS = _os.path.abspath('../data/grafica_arbol_decision')
_os.makedirs(_GRAFICAS, exist_ok=True)

def evaluar_split(X, y, test_size, split_label, resultados_globales,
                  save_dir=_GRAFICAS):
    """
    Entrena un DecisionTreeRegressor, calcula métricas y genera
    dos gráficos: residuos y predicciones vs. reales.

    Parámetros
    ----------
    X                   : array-like — variable predictora (fuerza)
    y                   : array-like — variable objetivo (poder)
    test_size           : float      — fracción de prueba (ej. 0.20)
    split_label         : str        — etiqueta del split para títulos/guardado
    resultados_globales : list       — donde se agregan los resultados del split
    save_dir            : str        — directorio donde guardar las gráficas
    """
    train_pct = int((1 - test_size) * 100)
    test_pct  = int(test_size * 100)

    # ── 1. Partición ──────────────────────────────────────────
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=42
    )

    print(f"\n{'='*58}")
    print(f"  {split_label} — {train_pct}% Train / {test_pct}% Test")
    print(f"{'='*58}")
    print(f"  Train: {X_train.shape[0]:,} muestras | "
          f"Test: {X_test.shape[0]:,} muestras")

    # ── 2. Entrenamiento ──────────────────────────────────────
    modelo = DecisionTreeRegressor(random_state=42)
    modelo.fit(X_train, y_train)

    # ── 3. Predicciones ───────────────────────────────────────
    y_pred = modelo.predict(X_test)

    # ── 4. Métricas ───────────────────────────────────────────
    r2   = r2_score(y_test, y_pred)
    mse  = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    mae  = mean_absolute_error(y_test, y_pred)

    print(f"\n  📊 MÉTRICAS ({split_label})")
    print(f"  {'─'*45}")
    print(f"  R² (Coef. Determinación)  : {r2:.4f} ({r2*100:.2f}%)")
    print(f"  → Interpreta: el modelo explica el {r2*100:.1f}% de la")
    print(f"    varianza del Poder del superhéroe.")
    print(f"  MSE (Error Cuadrático Med): {mse:.4f}")
    print(f"  → Penaliza errores grandes; sensible a outliers.")
    print(f"  RMSE (Raíz del MSE)       : {rmse:.4f} pts")
    print(f"  → Error promedio expresado en la misma unidad que Y.")
    print(f"  MAE (Error Absoluto Med)  : {mae:.4f} pts")
    print(f"  → Desviación promedio absoluta de las predicciones.")
    print(f"  Profundidad del árbol     : {modelo.get_depth()}")
    print(f"  Número de hojas           : {modelo.get_n_leaves()}")

    # Guardar para tabla consolidada
    resultados_globales.append({
        'Split'    : split_label,
        'Train%'   : train_pct,
        'Test%'    : test_pct,
        'R²'       : round(r2,   4),
        'MSE'      : round(mse,  4),
        'RMSE'     : round(rmse, 4),
        'MAE'      : round(mae,  4),
    })

    # ── 5. Gráficos ───────────────────────────────────────────
    residuos = y_test - y_pred
    slug = (split_label.replace('/', '-')
                       .replace(' ', '_')
                       .replace('(', '')
                       .replace(')', ''))

    fig, axes = plt.subplots(1, 2, figsize=(16, 6))
    fig.suptitle(
        f'DecisionTreeRegressor — {split_label} · Superhéroes',
        fontsize=14, fontweight='bold'
    )

    # Gráfico 1: Residuos vs Predichos + LOWESS
    orden_idx = np.argsort(y_pred)
    axes[0].scatter(y_pred, residuos,
                    alpha=0.35, s=14, color='#4ecdc4', label='Residuos')
    axes[0].axhline(0, color='#ff6b6b', linestyle='--',
                    linewidth=2, label='Línea ideal (error = 0)')
    lowess_res = sm.nonparametric.lowess(
        residuos[orden_idx], y_pred[orden_idx], frac=0.4
    )
    axes[0].plot(lowess_res[:, 0], lowess_res[:, 1],
                 color='#ffa07a', linewidth=2.5, label='Tendencia LOWESS')
    axes[0].set_title('Gráfico de Errores (Residuos)', fontweight='bold')
    axes[0].set_xlabel('Poder Predicho (pts)')
    axes[0].set_ylabel('Residuos (pts)')
    axes[0].legend()
    axes[0].grid(alpha=0.3)

    # Gráfico 2: Predicción vs Real
    minval = min(y_test.min(), y_pred.min())
    maxval = max(y_test.max(), y_pred.max())
    axes[1].scatter(y_test, y_pred,
                    alpha=0.30, s=14, color='#95e1d3', label='Predicciones')
    axes[1].plot([minval, maxval], [minval, maxval],
                 color='#ff6b6b', linewidth=2, linestyle='--',
                 label='Predicción perfecta')
    axes[1].set_title(
        f'Predicción vs. Valor Real  (R² = {r2:.4f})', fontweight='bold'
    )
    axes[1].set_xlabel('Poder Real (pts)')
    axes[1].set_ylabel('Poder Predicho (pts)')
    axes[1].legend()
    axes[1].grid(alpha=0.3)

    plt.tight_layout()
    fname = _os.path.join(save_dir, f'dt_{slug}.png')
    plt.savefig(fname, dpi=150, bbox_inches='tight')
    plt.show()
    print(f"  💾 Gráfico guardado → {fname}")

    return modelo, y_test, y_pred

# Contenedor de resultados para el análisis consolidado
resultados_globales = []
print("✅ Función evaluar_split definida correctamente.")
print(f"   Directorio de gráficas: {_GRAFICAS}")
```

### CELDA 5 — Split 1: 80% Entrenamiento / 20% Prueba

Partición estándar y referencia del análisis. La misma usada en el notebook 01,
lo que permite comparación directa entre la regresión lineal y el árbol de decisión.
Maximiza los datos de entrenamiento (2 828 muestras).

```python
# ── CELDA 5: Split 1 — 80/20 ─────────────────────────────────
modelo_s1, y_test_s1, y_pred_s1 = evaluar_split(
    X, y,
    test_size=0.20,
    split_label='Split 1 (80/20)',
    resultados_globales=resultados_globales,
)
```

### CELDA 6 — Split 2: 70% Entrenamiento / 30% Prueba

Reduce moderadamente el conjunto de entrenamiento (2 474 muestras).
Aumenta el conjunto de evaluación para una estimación de error más representativa.
Si las métricas se mantienen cercanas al split anterior, el modelo generaliza bien.

```python
# ── CELDA 6: Split 2 — 70/30 ─────────────────────────────────
modelo_s2, y_test_s2, y_pred_s2 = evaluar_split(
    X, y,
    test_size=0.30,
    split_label='Split 2 (70/30)',
    resultados_globales=resultados_globales,
)
```

### CELDA 7: Split 3 — 60% Entrenamiento / 40% Prueba

La partición más exigente. Con solo 60% para entrenar (2 121 muestras),
el árbol tiene menor capacidad de memorizar el entrenamiento.
La más útil para detectar sobreajuste: un árbol que memoriza caería notablemente aquí.

```python
# ── CELDA 7: Split 3 — 60/40 ─────────────────────────────────
modelo_s3, y_test_s3, y_pred_s3 = evaluar_split(
    X, y,
    test_size=0.40,
    split_label='Split 3 (60/40)',
    resultados_globales=resultados_globales,
)
```

### CELDA 8 — Análisis Consolidado (Summary)

Se realiza una comparación del rendimiento de los tres splits evaluados, analizando la estabilidad de las métricas de regresión y el comportamiento del modelo.

El análisis permite:
- Evaluar la capacidad de generalización del árbol.
- Identificar posibles indicios de sobreajuste o subajuste.
- Determinar si el modelo requiere ajustes adicionales o el uso de métodos más avanzados.

```python
# ── CELDA 8: Tabla comparativa de métricas ───────────────────
df_resumen = pd.DataFrame(resultados_globales)

print("=" * 65)
print("  TABLA COMPARATIVA — TRES SPLITS · ÁRBOL DE DECISIÓN")
print("  Variable predictora: fuerza  |  Variable objetivo: poder")
print("=" * 65)
print(df_resumen.to_string(index=False))
print("=" * 65)

print("\n" + "=" * 65)
print("  REFERENCIA — Regresión Lineal Múltiple (Notebook 01)")
print("  Variables: fuerza + inteligencia + combate  |  Split 80/20")
print("=" * 65)
print(f"  {'Split':<20} {'Train%':>7} {'Test%':>7} {'R²':>7} "
      f"{'MSE':>9} {'RMSE':>7} {'MAE':>7}")
print(f"  {'Reg. Lineal Múltiple':<20} {'80':>7} {'20':>7} "
      f"{'0.3944':>7} {'33.5811':>9} {'5.7949':>7} {'4.6598':>7}")
print("=" * 65)
```

### CELDA 9 — Gráfico comparativo de métricas por split

```python
# ── CELDA 9: Gráfico comparativo de métricas por split ───────
metricas = ['R²', 'RMSE', 'MAE', 'MSE']
colores  = ['#ff6b6b', '#4ecdc4', '#95e1d3', '#ffa07a']

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle(
    'Comparativa de Métricas — DecisionTreeRegressor · Superhéroes (3 Splits)',
    fontsize=14, fontweight='bold', y=1.01
)

for ax, (metrica, color) in zip(axes.flat, zip(metricas, colores)):
    valores = df_resumen[metrica].values
    barras  = ax.bar(
        df_resumen['Split'], valores,
        color=color, edgecolor='white', width=0.5
    )
    ax.set_title(metrica, fontweight='bold')
    ax.set_ylabel('Valor')
    ax.set_ylim(0, max(valores) * 1.2 + 1e-6)
    ax.grid(axis='y', alpha=0.3)
    ax.tick_params(axis='x', rotation=10)

    for bar, val in zip(barras, valores):
        ax.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height() + max(valores) * 0.03,
            f'{val:.4f}',
            ha='center', va='bottom',
            fontsize=10, fontweight='bold'
        )

    # Resaltar Split 1 (80/20) como referencia del notebook 01
    barras[0].set_edgecolor('#1f3c88')
    barras[0].set_linewidth(2)

plt.tight_layout()
fname_comp = '../data/grafica_arbol_decision/dt_comparativa_splits.png'
plt.savefig(fname_comp, dpi=150, bbox_inches='tight')
plt.show()
print(f"💾 Gráfico comparativo guardado → {fname_comp}")
print("\n📌 La barra con borde azul oscuro es el Split 1 (80/20), referencia del notebook 01.")
```

```python
### CELDA 10 — Análisis interpretativo y recomendación
```

```python
# ── CELDA 10: Análisis interpretativo y recomendación ────────
r2_vals   = df_resumen['R²'].values
rmse_vals = df_resumen['RMSE'].values
mae_vals  = df_resumen['MAE'].values

r2_rango   = r2_vals.max()   - r2_vals.min()
rmse_rango = rmse_vals.max() - rmse_vals.min()

print("=" * 65)
print("   ANÁLISIS CONSOLIDADO — DecisionTreeRegressor")
print("=" * 65)

print("""
1. VARIACIÓN DEL RENDIMIENTO ENTRE SPLITS
─────────────────────────────────────────""")
print(f"   R²   máx={r2_vals.max():.4f}  mín={r2_vals.min():.4f}  "
      f"rango={r2_rango:.4f}")
print(f"   RMSE máx={rmse_vals.max():.4f}  mín={rmse_vals.min():.4f}  "
      f"rango={rmse_rango:.4f}")

if r2_rango < 0.05:
    print("""
   ✅ El R² es estable entre los splits.
      El modelo presenta un comportamiento consistente y no depende
      fuertemente del tamaño del conjunto de entrenamiento.""")
else:
    print("""
   ⚠️  El R² varía significativamente entre splits.
      Esto indica sensibilidad al tamaño de los datos y posible
      inestabilidad del modelo.""")

print("""
2. INTERPRETACIÓN DEL PODER PREDICTIVO
──────────────────────────────────────
   • Los valores de R² (~0.15) indican que el modelo solo explica
     aproximadamente el 15% de la variabilidad del poder.

   • Esto sugiere que la variable fuerza, aunque es el mejor predictor
     individual, NO es suficiente para explicar completamente el poder.

   • El modelo presenta una capacidad predictiva débil, lo cual es
     consistente en todos los splits evaluados.
""")

print("""
3. INTERPRETACIÓN DE LOS ERRORES
────────────────────────────────
   • RMSE ≈ 6.8 – 7.0 → el error promedio de predicción es de ~7 puntos
     en la variable poder.

   • MAE ≈ 5.3 – 5.5 → en promedio, las predicciones se desvían unos
     5 puntos del valor real.

   • Estos valores indican una precisión moderada, pero no alta.

   • Si el gráfico de residuos muestra dispersión aleatoria → el modelo
     es coherente, aunque limitado en capacidad explicativa.
""")

best_split = df_resumen.loc[df_resumen['R²'].idxmax(), 'Split']
best_r2    = df_resumen['R²'].max()
best_rmse  = df_resumen.loc[df_resumen['R²'].idxmax(), 'RMSE']

print(f"   Mejor split: {best_split}  →  R²={best_r2:.4f}, RMSE={best_rmse:.4f}")

print("""
4. RECOMENDACIÓN
────────────────
   A) AJUSTE DEL MODELO:
      • Reducir la complejidad del árbol:
        - max_depth=3..6
        - min_samples_leaf=20..50

      • Esto puede mejorar la generalización y reducir overfitting.

   B) ANÁLISIS DE VARIABLES:
      • La variable fuerza es la más relevante individualmente.
      • Sin embargo, el bajo R² indica que el poder depende de
        múltiples factores no capturados completamente.

   C) MODELOS ALTERNATIVOS:
      • RandomForestRegressor → reduce varianza
      • GradientBoostingRegressor → mejora precisión
      • XGBoost → mejor desempeño en problemas complejos

   CONCLUSIÓN:
   El modelo basado en fuerza presenta un rendimiento consistente,
   pero con bajo poder explicativo. Esto sugiere que el fenómeno
   (poder) es más complejo y requiere múltiples variables o modelos
   más avanzados para ser predicho con mayor precisión.
""")

print("=" * 65)
```

### CELDA 11 —  Visualización del Árbol de Decisión

El árbol entrenado puede ser muy profundo (sin restricciones de `max_depth`).
Para entender el mecanismo de predicción se generan tres representaciones:

1. **Árbol real truncado** — primeros 4 niveles del `modelo_s1` (Split 80/20). Muestra los umbrales reales de `fuerza` aprendidos.
2. **Árbol pedagógico** — reentrenado con `max_depth=3` para mostrar con claridad la lógica completa: condición de corte · muestras · poder predicho en cada hoja.
3. **Reglas de decisión en texto** — lectura explícita de todos los caminos del árbol simplificado.

```python
# ── CELDA 11: Árbol real — Split 80/20 (primeros 4 niveles) ──
import os as _os

_GRAFICAS_ARBOL = _os.path.abspath('../data/grafica_arbol_decision')
_os.makedirs(_GRAFICAS_ARBOL, exist_ok=True)

fig, ax = plt.subplots(figsize=(22, 10))

plot_tree(
    modelo_s1,
    max_depth=4,
    feature_names=['Fuerza'],
    filled=True,
    rounded=True,
    impurity=False,
    precision=2,
    fontsize=9,
    ax=ax,
)

ax.set_title(
    'Árbol de Decisión — Split 80/20 (primeros 4 niveles de profundidad)\n'
    f'Profundidad total: {modelo_s1.get_depth()} | '
    f'Hojas totales: {modelo_s1.get_n_leaves()} | '
    f'R² = {r2_score(y_test_s1, y_pred_s1):.4f}',
    fontsize=13, fontweight='bold', pad=15
)

plt.tight_layout()
fname_real = _os.path.join(_GRAFICAS_ARBOL, 'dt_arbol_real_4niveles.png')
plt.savefig(fname_real, dpi=150, bbox_inches='tight')
plt.show()

print(f"💾 Árbol real guardado → {fname_real}")
print(f"\n📐 Información del árbol completo (sin restricción de profundidad):")
print(f"   Profundidad máxima : {modelo_s1.get_depth()}")
print(f"   Número de hojas    : {modelo_s1.get_n_leaves()}")
print(f"   Nodos totales      : {modelo_s1.tree_.node_count}")
print(f"\n   Un árbol de profundidad {modelo_s1.get_depth()} ha memorizado casi todos los")
print(f"   patrones del entrenamiento — riesgo de sobreajuste con datos más ruidosos.")
```

### CELDA 12 — Árbol pedagógico simplificado

```python
# ── CELDA 12: Árbol pedagógico simplificado (max_depth=3) ─────
from sklearn.model_selection import train_test_split as _tts

# Reentrenar con max_depth=3 sobre el mismo split 80/20 para consistencia
X_train_viz, X_test_viz, y_train_viz, y_test_viz = _tts(
    X, y, test_size=0.20, random_state=42
)

modelo_viz = DecisionTreeRegressor(max_depth=3, random_state=42)
modelo_viz.fit(X_train_viz, y_train_viz)

r2_viz  = r2_score(y_test_viz, modelo_viz.predict(X_test_viz))
r2_full = r2_score(y_test_s1, y_pred_s1)

print(f"📊 Comparación árbol completo vs árbol pedagógico (max_depth=3):")
print(f"   Árbol completo  (depth={modelo_s1.get_depth():>2}): "
      f"R² = {r2_full:.4f}  |  hojas = {modelo_s1.get_n_leaves()}")
print(f"   Árbol pedagógico (depth= 3): "
      f"R² = {r2_viz:.4f}  |  hojas = {modelo_viz.get_n_leaves()}")
print(f"   Pérdida de R² por simplificar: {r2_full - r2_viz:.4f}")

fig, ax = plt.subplots(figsize=(24, 12))

plot_tree(
    modelo_viz,
    max_depth=3,
    feature_names=['Fuerza (pts)'],
    filled=True,
    rounded=True,
    impurity=False,
    precision=2,
    fontsize=11,
    ax=ax,
)

ax.set_title(
    f'Árbol Pedagógico — max_depth=3  |  {modelo_viz.get_n_leaves()} hojas  |  R² = {r2_viz:.4f}\n'
    f'(vs árbol completo: depth={modelo_s1.get_depth()}, '
    f'{modelo_s1.get_n_leaves()} hojas, R²={r2_full:.4f})',
    fontsize=13, fontweight='bold', pad=15
)

plt.tight_layout()
fname_ped = _os.path.join(_GRAFICAS_ARBOL, 'dt_arbol_pedagogico_depth3.png')
plt.savefig(fname_ped, dpi=180, bbox_inches='tight')
plt.show()

print(f"\n💾 Árbol pedagógico guardado → {fname_ped}")
print(f"\n📊 Rendimiento del árbol simplificado (max_depth=3):")
print(f"   R² = {r2_viz:.4f}  (vs {r2_full:.4f} del árbol sin restricción)")
print(f"\n✅ Con solo 3 niveles de profundidad ({modelo_viz.get_n_leaves()} hojas),")
print(f"   el árbol alcanza R² = {r2_viz:.4f}, apenas {r2_full - r2_viz:.4f} por debajo")
print(f"   del árbol completo ({r2_full:.4f}).")
print(f"   Los primeros cortes de fuerza capturan la mayor parte")
print(f"   de la información predictiva del poder.")
```
